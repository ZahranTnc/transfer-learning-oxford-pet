```python
import os, json, csv, time, random, copy, shutil
import numpy as np
import torch, torch.nn as nn
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, random_split
from torchvision import models, transforms, datasets
from sklearn.metrics import f1_score, accuracy_score
```

```python
SEED, BATCH, NCLS, PATIENCE, EPOCHS = 42, 32, 10, 4, 25
dev = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
mean, std = [0.485,0.456,0.406], [0.229,0.224,0.225]
metrics = []
print(dev)
```

```python
raw = datasets.OxfordIIITPet('./data', 'trainval', download=True)
raw2 = datasets.OxfordIIITPet('./data', 'test', download=True)
cls = raw.classes
chosen = [c for c in cls if c[0].isupper()][:5] + [c for c in cls if c[0].islower()][:5]

PATH = 'pet10'
if not os.path.exists(PATH):
    for c in chosen: os.makedirs(f'{PATH}/{c}', exist_ok=True)
    for ds in [raw, raw2]:
        for img_path, lbl in zip(ds._images, ds._labels):
            name = cls[lbl]
            if name in chosen: shutil.copy(img_path, f'{PATH}/{name}/{os.path.basename(img_path)}')
print(chosen)
```

```python
train_tf = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomResizedCrop(224, scale=(0.8,1.0)),
    transforms.RandomHorizontalFlip(0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(0.2, 0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)])

eval_tf = transforms.Compose([
    transforms.Resize(256), transforms.CenterCrop(224),
    transforms.ToTensor(), transforms.Normalize(mean, std)])
```

```python
full = datasets.ImageFolder(PATH)
n = len(full)
ntr, nva = int(n*0.7), int(n*0.15)
g = torch.Generator().manual_seed(SEED)
s_tr, s_va, s_te = random_split(full, [ntr, nva, n-ntr-nva], generator=g)
print(n, len(s_tr), len(s_va), len(s_te))
```

```python
class Wrap(torch.utils.data.Dataset):
    def __init__(self, sub, tf): self.sub, self.tf = sub, tf
    def __len__(self): return len(self.sub)
    def __getitem__(self, i):
        img, y = self.sub[i]
        return self.tf(img), y

def loaders(ttf):
    return (DataLoader(Wrap(s_tr, ttf), BATCH, shuffle=True),
            DataLoader(Wrap(s_va, eval_tf), BATCH),
            DataLoader(Wrap(s_te, eval_tf), BATCH))
```

```python
def make(mode):
    w = None if mode == 'scratch' else models.ResNet18_Weights.IMAGENET1K_V1
    m = models.resnet18(weights=w)
    if mode in ('feature', 'finetune'):
        for p in m.parameters(): p.requires_grad = False
    m.fc = nn.Linear(m.fc.in_features, NCLS)
    if mode == 'finetune':
        for p in m.layer4.parameters(): p.requires_grad = True
    return m.to(dev)

def opt(m, mode):
    if mode == 'finetune':
        return torch.optim.AdamW([{'params': m.layer4.parameters(), 'lr': 1e-4},
                                  {'params': m.fc.parameters(), 'lr': 1e-3}], weight_decay=1e-2)
    g = m.fc.parameters() if mode == 'feature' else m.parameters()
    return torch.optim.AdamW(g, lr=1e-3, weight_decay=1e-2)
```

```python
lossf = nn.CrossEntropyLoss()

def ev(m, dl):
    m.eval(); s, ys, ps = 0, [], []
    with torch.no_grad():
        for x, y in dl:
            x, y = x.to(dev), y.to(dev)
            o = m(x); s += lossf(o, y).item() * len(x)
            ps += o.argmax(1).cpu().tolist(); ys += y.cpu().tolist()
    return s/len(dl.dataset), accuracy_score(ys, ps), f1_score(ys, ps, average='macro')
```

```python
def run(code, mode, ttf, lr):
    train, val, tst = loaders(ttf)
    m = make(mode); o = opt(m, mode)
    tp = sum(p.numel() for p in m.parameters() if p.requires_grad)
    print(f'\n{code} ({mode}) trainable={tp:,}')
    best, bs, wait, conv, h = 1e9, None, 0, 0, {'tl': [], 'vl': []}
    t0 = time.time()
    for e in range(1, EPOCHS+1):
        m.train(); tot = 0
        for x, y in train:
            x, y = x.to(dev), y.to(dev)
            o.zero_grad(); l = lossf(m(x), y); l.backward(); o.step()
            tot += l.item() * len(x)
        tl = tot/len(train.dataset)
        vl, va, _ = ev(m, val)
        h['tl'].append(tl); h['vl'].append(vl)
        print(f'  ep{e:02d} tl={tl:.3f} vl={vl:.3f} va={va:.3f}')
        if vl < best - 1e-4:
            best, bs, wait, conv = vl, copy.deepcopy(m.state_dict()), 0, e
        else:
            wait += 1
            if wait >= PATIENCE: break          # early stopping
    m.load_state_dict(bs)
    _, vacc, _ = ev(m, val)
    _, tacc, tf1 = ev(m, tst)
    metrics.append({'kondisi': code, 'strategi': mode, 'lr': lr, 'trainable': tp,
                    'epoch_konv': conv, 'durasi_s': round(time.time()-t0, 1),
                    'val_acc': round(vacc, 4), 'test_acc': round(tacc, 4), 'test_f1': round(tf1, 4)})
    return m, h
```

```python
m1, h1 = run('B1', 'scratch',  train_tf, '1e-3')
m2, h2 = run('B2', 'feature',  train_tf, '1e-3 fc')
m3, h3 = run('B3', 'finetune', train_tf, '1e-4 layer4 / 1e-3 fc')
```

```python
_, h4a = run('B4-noaug', 'finetune', eval_tf,  '1e-4/1e-3 no-aug')
_, h4b = run('B4-aug',   'finetune', train_tf, '1e-4/1e-3 aug')
print('gap no-aug:', round(h4a['tl'][-1]-h4a['vl'][-1], 3), '| gap aug:', round(h4b['tl'][-1]-h4b['vl'][-1], 3))
```

```python
train, val, tst = loaders(train_tf)
m5 = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)
for p in m5.parameters(): p.requires_grad = False
m5.classifier[1] = nn.Linear(m5.classifier[1].in_features, NCLS)
m5 = m5.to(dev)
o5 = torch.optim.AdamW(m5.classifier.parameters(), lr=1e-3, weight_decay=1e-2)
tp5 = sum(p.numel() for p in m5.parameters() if p.requires_grad)

best, bs, wait, conv = 1e9, None, 0, 0; t0 = time.time()
for e in range(1, EPOCHS+1):
    m5.train()
    for x, y in train:
        x, y = x.to(dev), y.to(dev)
        o5.zero_grad(); l = lossf(m5(x), y); l.backward(); o5.step()
    vl, va, _ = ev(m5, val)
    print(f'ep{e:02d} vl={vl:.3f} va={va:.3f}')
    if vl < best - 1e-4: best, bs, wait, conv = vl, copy.deepcopy(m5.state_dict()), 0, e
    else:
        wait += 1
        if wait >= PATIENCE: break
m5.load_state_dict(bs)
_, vacc, _ = ev(m5, val); _, tacc, tf1 = ev(m5, tst)
metrics.append({'kondisi': 'B5', 'strategi': 'effnet-b0', 'lr': '1e-3 head', 'trainable': tp5,
                'epoch_konv': conv, 'durasi_s': round(time.time()-t0, 1),
                'val_acc': round(vacc, 4), 'test_acc': round(tacc, 4), 'test_f1': round(tf1, 4)})
```

```python
json.dump(metrics, open('metrics_log.json', 'w'), indent=2)
w = csv.DictWriter(open('metrics_log.csv', 'w', newline=''), metrics[0].keys()); w.writeheader(); w.writerows(metrics)
import pandas as pd
pd.DataFrame(metrics)
```

```python
for name, h in [('B1', h1), ('B2', h2), ('B3', h3)]:
    plt.plot(h['vl'], label=name+' val')
    plt.plot(h['tl'], '--', label=name+' train')
plt.legend(); plt.title('train vs val loss'); plt.xlabel('epoch'); plt.ylabel('loss')
plt.savefig('loss_curves.png', dpi=120); plt.show()
```

```python
import torch.nn.functional as F
inv = transforms.Normalize([-m/s for m, s in zip(mean, std)], [1/s for s in std])
_, _, tst = loaders(train_tf)
m3.eval(); wrong = []
with torch.no_grad():
    for x, y in tst:
        p = F.softmax(m3(x.to(dev)), 1); cf, pr = p.max(1)
        for i in range(len(x)):
            if pr[i].item() != y[i].item():
                wrong.append((x[i].cpu(), y[i].item(), pr[i].item(), cf[i].item()))
        if len(wrong) >= 8: break

fig, ax = plt.subplots(2, 4, figsize=(16, 8))
for a, (img, gt, pr, cf) in zip(ax.ravel(), wrong[:8]):
    a.imshow(inv(img).clamp(0, 1).permute(1, 2, 0)); a.axis('off')
    a.set_title(f'GT {chosen[gt]}\nPred {chosen[pr]} {cf:.2f}', fontsize=9)
plt.tight_layout(); plt.savefig('failure_analysis.png', dpi=120); plt.show()
```
